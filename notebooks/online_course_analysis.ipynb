# Online Course Learning Analysis

A Pandas, NumPy and Matplotlib walkthrough of the online course dataset.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
df = pd.read_csv('../data/online_course_dataset.csv')
df.head()
```

## 1. Average score by course


```python
by_course = df.groupby('course').agg(
    students=('learner_id', 'count'),
    avg_watch_hours=('watch_hours', 'mean'),
    avg_quiz_score=('avg_quiz_score', 'mean'),
    avg_final_score=('final_score', 'mean')
).round(2).sort_values('avg_final_score', ascending=False)
by_course
```

## 2. Visualization: watch hours vs final score


```python
slope, intercept = np.polyfit(df['watch_hours'], df['final_score'], 1)
x = np.linspace(df['watch_hours'].min(), df['watch_hours'].max(), 100)

plt.figure(figsize=(9, 5))
plt.scatter(df['watch_hours'], df['final_score'], alpha=0.65)
plt.plot(x, slope * x + intercept, color='black', linewidth=1.5)
plt.title('Watch Hours vs Final Score')
plt.xlabel('Watch hours')
plt.ylabel('Final score')
plt.show()
```
